# 01 - Dataset Exploration

## 1.1 Check the Data Files

```python
from pathlib import Path

import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw"

expected_files = [
    "olist_customers_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "product_category_name_translation.csv"
]

for file in expected_files:
    path = DATA_PATH / file

    if path.exists():
        print(f"[OK] {file}")
    else:
        print(f"[MISSING] {file}")
```

## 1.2 Load The Data & Check The Structure

```python
datasets = {}

for file in expected_files:
    table_name = file.replace(".csv", "")

    df = pd.read_csv(DATA_PATH/file)

    datasets[table_name] = df

    print(

        f"{table_name}:"
        f"{df.shape[0]:,} rows,"
        f"{df.shape[1]} columns"

    )
```

```python
for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"TABLE: {name}")
    print(f"{'=' * 70}")

    display(df.head(5))

    print("\nData Types:")
    print(df.dtypes)
```

## 1.4 Data Quality Profiling

```python
quality_results = []

for name, df in datasets.items():

    quality_results.append({
        "Table": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Null Cells": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum(),
        "Memory (MB)": round(
            df.memory_usage(deep=True).sum() / 1024**2, 2
        )
    })


quality_df = pd.DataFrame(quality_results)

display(quality_df)
```

```python
for name, df in datasets.items():
    null_counts = df.isnull().sum()

    null_counts = null_counts[null_counts > 0]

    if not null_counts.empty:
        print(f"\n{name}")

        print(
            pd.DataFrame({
                "Null Count": null_counts,
                "Null Percentage": (
                    null_counts / len(df) * 100
                ).round(2)
            })
        )
```

## 1.5 Check Relations & Primary Keys

```python
primary_keys = {
    "olist_customers_dataset": ["customer_id"],
    "olist_orders_dataset": ["order_id"],
    "olist_order_items_dataset": ["order_id", "order_item_id"],
    "olist_order_payments_dataset": ["order_id", "payment_sequential"],
    "olist_order_reviews_dataset": ["review_id"],
    "olist_products_dataset": ["product_id"],
    "olist_sellers_dataset": ["seller_id"],
    "product_category_name_translation": ["product_category_name"]
}

for table, keys in primary_keys.items():

    df = datasets[table]

    duplicate_keys = df.duplicated(
        subset=keys
    ).sum()

    print(
        f"{table}: "
        f"Duplicate Keys = {duplicate_keys:,}"
    )
```

## 1.6 Save The First Report

```python
REPORT_PATH = PROJECT_ROOT / "docs"

REPORT_PATH.mkdir(
    parents=True,
    exist_ok=True
)

quality_df.to_csv(
    REPORT_PATH / "initial_data_quality.csv",
    index=False
)

print("Data quality report exported successfully.")
```

## 1.7 Check Anomaly

### 1.7.1 Test : Duplicate Reviews

```python
reviews = datasets["olist_order_reviews_dataset"]

# find duplicated review IDs
duplicated_reviews = reviews[
    reviews.duplicated(
        subset=["review_id"],
        keep=False
    )
].sort_values("review_id")

print("Duplicated review_id rows:",
      len(duplicated_reviews))

display(duplicated_reviews.head(20))

# test composite key
print(
    "Duplicate review_id + order_id:",
    reviews.duplicated(
        subset=["review_id", "order_id"]
    ).sum()
)

# number of distinct orders per review_id
display(
    duplicated_reviews.groupby("review_id")
    ["order_id"]
    .nunique()
    .value_counts()
)
```

### 1.7.2 Test : Orders NULL values relations with Order Status

```python
orders = datasets["olist_orders_dataset"]

null_analysis = (
    orders.groupby(
        "order_status"
    ).agg(
        Total_Orders=("order_id", "count"),
        Missing_Approval=("order_approved_at",
                          lambda x: x.isna().sum()),
        Missing_Carrier=("order_delivered_carrier_date",
                         lambda x: x.isna().sum()),
        Missing_Delivery=("order_delivered_customer_date",
                         lambda x: x.isna().sum())
    )
)

display(null_analysis)
```

### 1.7.3 Test : Geolocation

```python
geo = datasets["olist_geolocation_dataset"]

print("Total rows:", len(geo))

print(
    "Exact duplicate rows:",
    geo.duplicated().sum()
)

print(
    "Unique ZIP codes:",
    geo["geolocation_zip_code_prefix"].nunique()
)

print(
    "Unique ZIP + coordinates:",
    geo.drop_duplicates(
        subset=[
            "geolocation_zip_code_prefix",
            "geolocation_lat",
            "geolocation_lng"
        ]
    ).shape[0]
)

print(
    "Duplicate ZIP codes:",
    geo.duplicated(
        subset=["geolocation_zip_code_prefix"]
    ).sum()
)
```

### 1.7.4 Test : Products NULLs

```python
products = datasets["olist_products_dataset"]

missing_product_info = products[
    products["product_category_name"].isna()
]

print("Products with missing category:",
      len(missing_product_info))

display(
    missing_product_info.head(10)
)

print(
    "Missing category and missing name length:",
    (
        products["product_category_name"].isna()
        & products["product_name_lenght"].isna()
    ).sum()
)
```

## 1.8 Check duplicate Review Id

```python
reviews = datasets["olist_order_reviews_dataset"]

duplicate_ids = reviews[
    reviews.duplicated(
        subset=["review_id"],
        keep=False
    )
]

# check whether all review attributes
# except order_id are identical per review_id
review_consistency = (
    duplicate_ids.groupby("review_id").agg(
        Order_Count=("order_id", "nunique"),
        Score_Count=("review_score", lambda s: s.nunique(dropna=False)),
        Title_Count=("review_comment_title", lambda s: s.nunique(dropna=False)),
        Message_Count=("review_comment_message", lambda s: s.nunique(dropna=False)),
        Creation_Date_Count=("review_creation_date", lambda s: s.nunique(dropna=False)),
        Answer_Date_Count=("review_answer_timestamp", lambda s: s.nunique(dropna=False))
    )
)

display(
    review_consistency
    .describe()
)

print(
    "Review IDs with different scores:",
    (review_consistency["Score_Count"] > 1).sum()
)

print(
    "Review IDs with different messages:",
    (review_consistency["Message_Count"] > 1).sum()
)

print(
    "Review IDs with different timestamps:",
    (
        (review_consistency["Creation_Date_Count"] > 1)
        |
        (review_consistency["Answer_Date_Count"] > 1)
    ).sum()
)
```

```python
# check whether the same review_id is related to the same customer or different customers.
review_orders = (
    reviews.merge(
            datasets["olist_orders_dataset"][
                    ["order_id","customer_id"]
                ],
                on = "order_id",
                how = "left",
                validate="many_to_one"
        )
        .merge(
            datasets["olist_customers_dataset"][
                ["customer_id", "customer_unique_id"]
            ],
            on = "customer_id",
            how = "left",
            validate = "many_to_one"
        )
)

review_customer_check = (
    review_orders[
        review_orders["review_id"].isin(
            duplicate_ids["review_id"]
        )
    ]
    .groupby("review_id")
    .agg(
        Orders = ("order_id", "nunique"),
        Unique_Customers = ("customer_unique_id", "nunique")
    )
)

display(review_customer_check.value_counts())
```
